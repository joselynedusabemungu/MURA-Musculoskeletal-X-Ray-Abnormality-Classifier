# MURA v1.1: Lightweight Musculoskeletal X-ray Classification

**Scope:** bone and musculoskeletal anatomy; plain X-ray radiographs; binary image classification; target classes **Normal** and **Abnormal**. Dataset: [MURA v1.1](https://stanfordmlgroup.github.io/competitions/mura/), accessed 2026-09-15. The real-world question is whether a musculoskeletal radiographic image resembles a normal or abnormal labelled study. This is a coursework prototype for ML learning and is not a diagnostic tool.

## A1. Problem Definition and Clinical Framing

MURA labels studies as normal or abnormal across elbow, finger, forearm, hand, humerus, shoulder, and wrist radiographs. This notebook predicts an image-level label for compatibility with a one-image Streamlit upload. A positive/abnormal prediction is an estimated dataset class, not a diagnosis. The model should only receive 2D PNG/JPEG radiographs visually similar to MURA.

## A3. Data Preparation and Leakage-Safe Splitting

Images are converted to grayscale, resized to 128×128, and represented by grayscale pixel features. No image is augmented. The official MURA training directory is split by patient ID into train and validation. The official validation directory is held out as test. Assertions verify patient disjointness.

```python
train_df,val_df,test_df=make_splits(train_all,valid_all)
if QUICK_RUN:
    train_df=train_df.groupby('class_name',group_keys=False).head(1500).reset_index(drop=True)
    val_df=val_df.groupby('class_name',group_keys=False).head(400).reset_index(drop=True)
    test_df=test_df.groupby('class_name',group_keys=False).head(500).reset_index(drop=True)
print({k:len(v) for k,v in [('train',train_df),('validation',val_df),('test',test_df)]})
assert set(train_df.patient_id).isdisjoint(val_df.patient_id) and set(train_df.patient_id).isdisjoint(test_df.patient_id) and set(val_df.patient_id).isdisjoint(test_df.patient_id)
print('Leakage check passed: patient IDs are disjoint.')
```

## A4. Model Architecture and Justification

This notebook uses a lightweight grayscale-pixel plus logistic-regression pipeline. Resized grayscale pixels provide a transparent CPU baseline that is easy to install and reproduce. Logistic regression outputs a probability for each class. This is appropriate for binary classification and is considerably easier to install and run than a deep-learning framework.

```python
import numpy as np

def safe_make_features(frame, name):
    features = []
    labels = []
    paths = []
    skipped = []

    total = len(frame)

    for i, row in enumerate(frame.itertuples(index=False), start=1):
        try:
            features.append(extract_features(row.path))
            labels.append(row.label)
            paths.append(row.path)
        except Exception as error:
            skipped.append((row.path, str(error)))

        if i % 500 == 0 or i == total:
            print(f"{name}: processed {i}/{total}")

    print(f"{name}: readable images = {len(features)}")
    print(f"{name}: skipped images = {len(skipped)}")

    if skipped:
        print("First skipped image:", skipped[0][0])

    if not features:
        raise RuntimeError(f"No readable images found for {name}.")

    return (
        np.asarray(features, dtype=np.float32),
        np.asarray(labels),
        np.asarray(paths),
    )


X_train, y_train, train_paths = safe_make_features(train_df, "Training")
X_val, y_val, val_paths = safe_make_features(val_df, "Validation")
X_test, y_test, test_paths = safe_make_features(test_df, "Test")

print("Training feature shape:", X_train.shape)
print("Validation feature shape:", X_val.shape)
print("Test feature shape:", X_test.shape)

model = build_model()
model.fit(X_train, y_train)

print("Model fitted successfully.")
```

```python
from pathlib import Path
import joblib

PROJECT_ROOT = (
    Path.cwd().resolve().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd().resolve()
)

MODEL_DIR = PROJECT_ROOT / "models"
MODEL_DIR.mkdir(exist_ok=True)

metadata = {
    "features": "128x128 grayscale pixels",
    "image_size": [128, 128],
    "classes": ["Normal", "Abnormal"],
    "classifier": "balanced LogisticRegression",
    "seed": 42,
}

model_path = MODEL_DIR / "mura_grayscale_logistic.joblib"

joblib.dump(
    {
        "model": model,
        "metadata": metadata,
    },
    model_path,
)

print("Model exported successfully")
print("Path:", model_path)
print("Exists:", model_path.exists())
```

## A5. Training Process

The classifier uses standardized grayscale pixel features and balanced logistic regression with `max_iter=1000`. The validation curve below shows performance as the amount of training data increases. This is a diagnostic learning curve rather than neural-network epoch history.

```python
from sklearn.metrics import accuracy_score
from sklearn.model_selection import learning_curve
tr_sizes=np.linspace(.1,1.0,5)
from sklearn.base import clone
rows=[]
for frac in tr_sizes:
    n=max(20,int(len(X_train)*frac)); m=clone(model); m.fit(X_train[:n],y_train[:n]); rows.append({'fraction':frac,'train_accuracy':m.score(X_train[:n],y_train[:n]),'validation_accuracy':m.score(X_val,y_val)})
learning=pd.DataFrame(rows); display(learning)
learning.plot(x='fraction',y=['train_accuracy','validation_accuracy'],marker='o',ylim=(0,1),title='Learning curve'); plt.tight_layout(); plt.savefig(OUTPUTS/'learning_curve.png',dpi=160); plt.show()
```

## A6. Evaluation and Results

The held-out test results include accuracy, precision, recall, F1, ROC-AUC, a confusion matrix, ROC and precision-recall curves, and correct/incorrect image examples. The numerical results below are generated only after the notebook is run on the actual MURA files.

```python
from sklearn.metrics import classification_report,confusion_matrix,roc_auc_score,roc_curve,precision_recall_curve,average_precision_score
prob=model.predict_proba(X_test)[:,1]; pred=(prob>=.5).astype(int)
print(classification_report(y_test,pred,target_names=CLASS_NAMES,zero_division=0))
metrics={'accuracy':float(accuracy_score(y_test,pred)),'roc_auc':float(roc_auc_score(y_test,prob)),'average_precision':float(average_precision_score(y_test,prob))}; print(metrics); json.dump(metrics,open(OUTPUTS/'metrics.json','w'),indent=2)
cm=confusion_matrix(y_test,pred); sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',xticklabels=CLASS_NAMES,yticklabels=CLASS_NAMES); plt.xlabel('Predicted'); plt.ylabel('True'); plt.title('Confusion matrix'); plt.tight_layout(); plt.savefig(OUTPUTS/'confusion_matrix.png',dpi=160); plt.show()
fpr,tpr,_=roc_curve(y_test,prob); precision,recall,_=precision_recall_curve(y_test,prob); fig,ax=plt.subplots(1,2,figsize=(11,4)); ax[0].plot(fpr,tpr,label=f"AUC={metrics['roc_auc']:.3f}"); ax[0].plot([0,1],[0,1],'--',color='gray'); ax[0].legend(); ax[0].set_title('ROC curve'); ax[1].plot(recall,precision,label=f"AP={metrics['average_precision']:.3f}"); ax[1].legend(); ax[1].set_title('Precision-recall curve'); plt.tight_layout(); plt.savefig(OUTPUTS/'roc_pr_curves.png',dpi=160); plt.show()
```

```python
def examples(idx,title):
    fig,axes=plt.subplots(1,len(idx),figsize=(5*len(idx),4)); axes=np.atleast_1d(axes)
    for ax,i in zip(axes,idx):
        with Image.open(test_paths[i]) as im: ax.imshow(im.convert('L'),cmap='gray')
        conf=prob[i] if pred[i] else 1-prob[i]; ax.set_title(f'True: {CLASS_NAMES[y_test[i]]}\nPred: {CLASS_NAMES[pred[i]]} ({conf:.1%})'); ax.axis('off')
    fig.suptitle(title); plt.tight_layout(); plt.show()
correct=np.where(pred==y_test)[0]; wrong=np.where(pred!=y_test)[0]
if len(correct): examples(correct[:min(3,len(correct))],'Correct predictions')
if len(wrong): examples(wrong[:min(3,len(wrong))],'Incorrect predictions')
```

## A8. Limitations, Ethics, and Next Steps

HOG is a baseline rather than a state-of-the-art medical imaging model. It may miss subtle findings, depends on acquisition similarity, and inherits MURA label noise and population limitations. False negatives and false positives have different clinical consequences. Real-world use would require external validation, calibration, subgroup analysis, prospective testing, privacy controls, regulatory review, and qualified clinical oversight.
